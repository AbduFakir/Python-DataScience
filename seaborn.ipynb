```python
import seaborn as sns
import pandas as pd

df = pd.read_csv('/content/class_ipl_22-26_dataset.csv')
df.tail(74)
```

```python
#Relational Plot

sns.relplot(
    data = df.tail(74),
    x = 'team1_runs',
    y = 'team2_runs',
    size = 'win_by_runs',
    kind='scatter'
)
```

```python
#Scatterplot

sns.scatterplot(
    x='team1_runs',
    y='team2_runs',
    hue='winner',
    data=df.tail(74)
)
```

```python
sns.scatterplot(
    x='team1_runs',
    y='team1_wickets',
    data=df.tail(74)
)
```

```python
#count Plot

sns.countplot(
    x='winner',
    data=df.tail(74)
).tick_params(axis='x',rotation=90)
```

```python
#Barplot

sns.barplot(
    x='season',
    y='team1_runs',
    data=df.tail(74),
    errorbar=None,
).tick_params(axis='x',rotation=90)
```

```python
#distribution plot

sns.displot(
    df['team2_runs'].tail(74),
    kde=True
)
```

```python
#histogram

sns.histplot(
    df['winner'],
    kde=True,
).tick_params(axis='x',rotation=90)
```

```python
#Boxplot

sns.boxplot(
    y=df['team1_runs'].tail(74)
)
```

```python
df.tail(74).describe()
```

```python
#pairplot

sns.pairplot(
    df[['team1_runs','team2_runs']].tail(74)
)
```

```python
#Correlation Matrix

correlation = df.corr(numeric_only=True)
correlation
```

```python
#Heatmap

sns.heatmap(
    correlation,
    annot=True,

)
```

```python
#KDE plot

sns.kdeplot(
    df['team1_runs'].tail(74),
    fill=True,
    bw_adjust=0.5
)
```
